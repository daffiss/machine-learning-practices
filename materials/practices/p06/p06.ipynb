### Task 1: KFold vs StratifiedKFold on an imbalanced problem

**Experiment.** Build a small imbalanced classification set (200 samples, ~3% positives) and inspect how many positive samples land in each fold under plain KFold and under StratifiedKFold.

```python
X, y = make_classification(n_samples=200, n_features=12, n_informative=6, n_redundant=2,
                           weights=[0.97, 0.03], n_clusters_per_class=1,
                           flip_y=0.02, random_state=1)
print("Overall positive rate: {:.3f}".format(y.mean()))

rows = []
for name, splitter in [("KFold", KFold(5, shuffle=True, random_state=42)),
                       ("StratifiedKFold", StratifiedKFold(5, shuffle=True, random_state=42))]:
    for fold, (tr, te) in enumerate(splitter.split(X, y)):
        rows.append({"splitter": name, "fold": fold,
                     "n_test": len(te),
                     "pos_in_train": int((y[tr] == 1).sum()),
                     "pos_in_test": int((y[te] == 1).sum())})
fold_stats = pd.DataFrame(rows)
print(fold_stats.to_string(index=False))
```

**Decision.** Which splitter would you use for this data, and why?


**Your answer:** I would choose StratifiedKFold because it approximately preserves the class proportions in each fold. In this imbalanced dataset, plain KFold distributes positive samples unevenly, and one fold contains no positive samples. StratifiedKFold provides a more balanced distribution of positive examples across folds, making model evaluation more reliable.

**Evidence.** What exactly in the table above is the evidence for your decision?


**Your answer:** The evidence is in the pos_in_test column. With KFold, the number of positive samples varies from 0 to 3 across folds, while StratifiedKFold distributes them more evenly, with 1 or 2 positive samples per fold. This makes StratifiedKFold more suitable for evaluating a model on this imbalanced dataset.

**Interpretation.** What is the practical consequence of a fold with zero (or one) positive samples, both for training and for evaluating the model?


**Your answer:** A fold with zero positive samples makes evaluation unreliable because we cannot properly assess how well the model identifies the positive class. With only one positive sample, metrics such as recall can change drastically depending on a single prediction. If the training fold contains no positive samples, the model may fail to learn the positive class. Therefore, StratifiedKFold is preferable for this imbalanced dataset.

**Counterfactual.** Predict what changes if (a) the number of folds is increased to 10, or (b) the positive rate drops to 0.5%. Which splitter becomes even more necessary?


**Your answer:** **(a)** Increasing the number of folds to 10 makes each validation fold smaller. ### Task 2: Time-ordered data - KFold, GroupKFold, TimeSeriesSplit

**Experiment.** Generate a synthetic series y_t = sin(2*pi*t/50) + 0.02*t + noise, with features [t/N, (t/N)^2]. Compare three validation schemes:

- KFold with shuffling (ignores order);
- GroupKFold over chunks of 20 consecutive points (respects grouping, ignores order);
- TimeSeriesSplit (expanding window, strictly past-to-future).

```python
N = 600
t = np.arange(N)
rng = np.random.RandomState(42)
y_series = np.sin(2 * np.pi * t / 50) + 0.02 * t + rng.normal(0, 0.08, N)
X_series = np.column_stack([t / N, (t / N) ** 2])
groups = t // 20

fig, ax = plt.subplots(figsize=(9, 3))
ax.plot(t, y_series, lw=1)
ax.set_xlabel("time index t")
ax.set_ylabel("y_t")
ax.set_title("Synthetic time series: trend + seasonal component + noise")
plt.tight_layout()
```

```python
schemes = {
    "KFold (shuffled)": KFold(5, shuffle=True, random_state=42),
    "GroupKFold (time chunks)": GroupKFold(n_splits=5),
    "TimeSeriesSplit": TimeSeriesSplit(n_splits=5),
}
cv_results = {}
for name, splitter in schemes.items():
    if name.startswith("GroupKFold"):
        cv_results[name] = cross_val_score(Ridge(), X_series, y_series, cv=splitter,
                                           groups=groups, scoring="r2")
    else:
        cv_results[name] = cross_val_score(Ridge(), X_series, y_series, cv=splitter, scoring="r2")

summary = pd.DataFrame(cv_results).T
summary.columns = ["fold {}".format(i + 1) for i in range(5)]
summary["mean R2"] = summary.mean(axis=1)
print(summary.round(3).to_string())

fig, ax = plt.subplots(figsize=(6.5, 3.5))
means = [v.mean() for v in cv_results.values()]
ax.bar(cv_results.keys(), means, color=["#4C72B0", "#DD8452", "#C44E52"])
ax.axhline(0, color="k", lw=0.8)
ax.set_ylabel("mean CV R2")
ax.set_title("Mean CV R2 by validation scheme (Ridge on [t, t^2])")
plt.tight_layout()
```

**Decision.** Which scheme gives the honest estimate of how the model will behave on future data, and why?


**Your answer:** I would choose TimeSeriesSplit because it preserves the chronological order of the data. The model is trained on past observations and evaluated on future observations, which best reflects how it would perform in practice. KFold may train on observations that occur after the validation period, while GroupKFold respects group boundaries but does not guarantee chronological order.


**Interpretation.** Why does the shuffled KFold look so good (R2 near 0.95) while TimeSeriesSplit collapses to a negative R2? Where does the "extra" information come from?


**Your answer:** Shuffled KFold gives an overly optimistic R² because the model can train on observations from later time periods and use patterns from across the entire series when predicting validation points. TimeSeriesSplit evaluates the model on future observations that come after the training data. Since the features capture the trend but not the sinusoidal pattern directly, the model struggles to extrapolate to future data, resulting in a negative R²(короче проблема в моделі, а не даних). This shows that the model may not generalize well to future observations.

**Counterfactual.** Suppose we add sin(2*pi*t/50) and cos(2*pi*t/50) as explicit features. Predict how the TimeSeriesSplit score would change, and explain the mechanism.


**Your answer:** Adding sin(2*pi*t/50) and cos(2*pi*t/50) as features should improve the TimeSeriesSplit score. The original features mainly capture the time trend but do not explicitly represent the periodic pattern. The new features allow the model to learn the sinusoidal component and apply it to future time points, improving its ability to generalize. Therefore, R² should increase, although noise may prevent it from reaching 1.

### Task 3: GridSearchCV inside a Pipeline vs leaking preprocessing

**Experiment.** Use a wide synthetic set (120 samples, 200 features, only 4 informative). Tune LogisticRegression C in two ways:

- Leaky: StandardScaler + SelectKBest are fit on the **full** dataset, then GridSearchCV runs on the already transformed matrix (scaling and feature selection never see the CV folds).
- Correct: StandardScaler + SelectKBest + LogisticRegression are wrapped in a Pipeline and GridSearchCV refits the whole pipeline inside every training fold.

```python
X, y = make_classification(n_samples=120, n_features=200, n_informative=4, n_redundant=10,
                           n_clusters_per_class=1, flip_y=0.12, random_state=42)

cv5 = StratifiedKFold(5, shuffle=True, random_state=42)

# --- LEAKY: preprocessing fit once on the full dataset ---
scaler_full = StandardScaler().fit(X)
X_scaled = scaler_full.transform(X)
selector_full = SelectKBest(f_classif, k=25).fit(X_scaled, y)
X_selected = selector_full.transform(X_scaled)
gs_leaky = GridSearchCV(LogisticRegression(max_iter=3000), {"C": [0.01, 0.1, 1.0, 10.0]},
                        cv=cv5).fit(X_selected, y)

# --- CORRECT: preprocessing inside the pipeline ---
pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("selector", SelectKBest(f_classif, k=25)),
    ("lr", LogisticRegression(max_iter=3000)),
])
gs_correct = GridSearchCV(pipe, {"lr__C": [0.01, 0.1, 1.0, 10.0]}, cv=cv5).fit(X, y)

print("LEAKY   best_score_ = {:.4f}   best C = {}".format(gs_leaky.best_score_, gs_leaky.best_params_))
print("CORRECT best_score_ = {:.4f}   best C = {}".format(gs_correct.best_score_, gs_correct.best_params_))
```

```python
fig, ax = plt.subplots(figsize=(5.5, 3.5))
ax.bar(["leaky (preprocessing on full data)", "correct (pipeline inside CV)"],
       [gs_leaky.best_score_, gs_correct.best_score_],
       color=["#C44E52", "#55A868"])
ax.set_ylim(0.5, 1.0)
ax.set_ylabel("best CV score")
ax.set_title("GridSearchCV: leaking vs correct preprocessing")
for i, v in enumerate([gs_leaky.best_score_, gs_correct.best_score_]):
    ax.text(i, v + 0.01, "{:.3f}".format(v), ha="center")
plt.tight_layout()
```

**Decision.** Which number would you report in a paper, and why is the other one optimistic?


**Your answer:** Я б репортнула CORRECT best score = 0.7667, оскільки в цьому випадку всі етапи обробки даних і навчання моделі виконуються всередині `Pipeline`. Для кожного фолду `StandardScaler` і `SelectKBest` навчаються лише на тренувальній частині, а потім застосовуються до валідаційної. Таким чином, валідаційні дані не впливають на попередню обробку.

Результат LEAKY = 0.8333 є занадто оптимістичним, оскільки `StandardScaler` і `SelectKBest` навчаються на всьому датасеті ще до крос-валідації. Зокрема, `SelectKBest` може вибрати ознаки, орієнтуючись також на зв’язок із цільовою змінною у валідаційних фолдах. Через це модель отримує інформацію, яка мала б залишатися невідомою під час оцінювання, і результат крос-валідації може бути завищеним.

## Discussion Questions (15 min)

1. Medical data contain several measurements per patient, and patients must never appear in both train and validation folds. Which splitter do you use, and why does plain KFold overstate performance?

I’d use GroupKFold with patient ID as the group, so all measurements of one patient land in the same fold. Plain KFold splits rows as if they were independent, so measurements from the same patient end up in both train and validation. The model can then recognize patient-specific patterns instead of learning something general, which is data leakage and makes the validation score overly optimistic.

2. Why is it not enough to shuffle rows for time-series data? What structure does the shuffle destroy, and which splitter respects it?

Time series data has temporal dependence: neighboring points are correlated and there are trends and seasonality. Shuffling destroys this order, so the model can train on future data to predict the past, and near-identical neighboring points end up in both train and validation, which inflates the score. TimeSeriesSplit respects the order: each fold trains on earlier data and validates on a later block, with the training window growing over time.

3. GridSearchCV reports the best score from the tuning folds. Why is that score an optimistic estimate for a tuned model, and what protocol gives an unbiased estimate (nested CV)?

GridSearchCV's best score can be optimistic because we select the best hyperparameters based on the same cross-validation results that we use to report performance. The best score may be inflated by random variation (selection bias).

Nested cross-validation solves this by using an inner CV loop for hyperparameter tuning and an outer CV loop for evaluation. The outer validation folds are not used during tuning, so their scores provide a less biased estimate of the performance of the entire model-selection process on unseen data.

4. Some preprocessing leaks are dangerous only when they see the target (feature selection, oversampling), others are dangerous even when unsupervised (scaling or imputation fit on full data when folds have different distributions). Give one example of each from today's experiments.

One target-dependent leakage example is fitting SelectKBest on the full dataset before cross-validation. Since feature selection uses the target labels, information from the validation folds influences which features are selected.

An unsupervised leakage example is fitting StandardScaler on the full dataset before cross-validation. Even without using target labels, the scaler learns the mean and standard deviation from the validation data. Both preprocessing steps should be fitted only on the training portion of each fold, for example by using a Pipeline.

5. A colleague says "I only looked at the test set once". Why is even a single use of the test set for a decision a problem?

Even a single use of the test set for a decision can introduce bias because the test results influence model selection or development. The test set is supposed to provide an independent estimate of performance on unseen data. If we use its results to change the model, it is no longer a fully independent evaluation.

6. When, if ever, is it legitimate to fit preprocessing on the full dataset before splitting?

Generally, fitting data-dependent preprocessing on the full dataset before splitting is not legitimate because information from the test set can leak into the training process. Preprocessing such as scaling, imputation, or feature selection should be fitted only on the training data, preferably inside a Pipeline.

A fixed, predefined transformation that does not learn from the dataset, such as applying a logarithm, can usually be performed before splitting. After model selection and evaluation are complete, we can also fit the final model and preprocessing on all available data for deployment, but we must retain an independent evaluation score.

## Takeaways

- Choose the validation scheme that matches the data-generating process: i.i.d. -> StratifiedKFold, grouped -> GroupKFold, time-ordered -> TimeSeriesSplit.
- Plain KFold on imbalanced data can produce folds without any positive samples, distorting both training and evaluation.
- Any preprocessing that uses validation-fold information (statistics or labels) before the split is leakage; the CV score becomes optimistic and can mislead model selection.
- GridSearchCV around a Pipeline refits every preprocessing step inside each training fold - this is the default way to tune a model.
- An optimistic CV score is not harmless: it creates wrong expectations for deployment and can select the wrong model.
- Nested CV (or a held-out test set used exactly once) is the protocol for an unbiased estimate of a tuned model.
